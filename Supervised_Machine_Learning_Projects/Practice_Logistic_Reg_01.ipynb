```python
import pandas as pd 
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("employee_turnover.csv")
```

```python
# NOW WE NEED TO CONVERT THE DATA INTO X AND Y

X = df.drop(columns = ['Employee_Turnover'])
y = df['Employee_Turnover']
```

```python
X.head()
```

```python
# Now We Split the data into train and test data 
from sklearn.model_selection import train_test_split

X_train, x_test, y_train, y_test = train_test_split(
    X, y, random_state = 42, test_size = 0.2
)
```

```python
# X_train # to see the data was split or not

# Now Import the model to train the model

from sklearn.linear_model import LogisticRegression
```

```python
# Train the model 

model = LogisticRegression()

model.fit(X_train, y_train)
```

```python
# Predict the output fron the data and check the accuracy score 
from sklearn.metrics import accuracy_score

y_pred = model.predict(x_test)

print(accuracy_score(y_pred, y_test))
```
